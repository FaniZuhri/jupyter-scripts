```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

# Parameter sinyal
Fs = 1000  # Frekuensi sampling (Hz)
T = 1.0 / Fs  # Periode sampling
N = 1024  # Jumlah sampel
t = np.linspace(0.0, N * T, N, endpoint=False)

# Sinyal individu
signal_50Hz = np.sin(2.0 * np.pi * 50 * t)
signal_120Hz = 0.5 * np.sin(2.0 * np.pi * 120 * t)
combined_signal = signal_50Hz + signal_120Hz

# Fourier Transform
yf = fft(combined_signal)
xf = fftfreq(N, T)

# --- Plot semua sinyal dalam satu grafik waktu ---
plt.figure(figsize=(12, 5))
plt.plot(t, signal_50Hz, label="50 Hz")
plt.plot(t, signal_120Hz, label="120 Hz")
plt.plot(t, combined_signal, label="Gabungan", linestyle="--", linewidth=2)
plt.title("Sinyal Waktu - 50 Hz, 120 Hz, dan Gabungannya")
plt.xlabel("Waktu (s)")
plt.ylabel("Amplitudo")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# --- Plot spektrum Fourier ---
plt.figure(figsize=(12, 4))
plt.plot(xf[: N // 2], 2.0 / N * np.abs(yf[: N // 2]))
plt.title("Magnitudo Fourier Transform - Gabungan 50 Hz dan 120 Hz")
plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Magnitudo")
plt.grid(True)
plt.tight_layout()
plt.show()
```
