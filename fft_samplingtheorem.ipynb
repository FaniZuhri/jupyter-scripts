```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

# Parameter sinyal
Fs = 1000  # Frekuensi sampling
T = 1 / Fs
N = 1024
t = np.linspace(0.0, N * T, N, endpoint=False)

# Sinyal: kombinasi 60 Hz dan 180 Hz
signal = np.sin(2 * np.pi * 60 * t) + 0.5 * np.sin(2 * np.pi * 180 * t)

# FFT
yf = fft(signal)
xf = fftfreq(N, T)

# Plot sinyal waktu
plt.figure(figsize=(12, 4))
plt.plot(t, signal)
plt.title("Sinyal Waktu - Kombinasi 60 Hz dan 180 Hz")
plt.xlabel("Waktu (s)")
plt.ylabel("Amplitudo")
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot FFT
plt.figure(figsize=(12, 4))
plt.plot(xf[: N // 2], 2.0 / N * np.abs(yf[: N // 2]))
plt.title("Magnitudo FFT")
plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Magnitudo")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sinyal asli: 30 Hz
f_signal = 30
t_cont = np.linspace(0, 1, 1000, endpoint=False)
signal_cont = np.sin(2 * np.pi * f_signal * t_cont)

# Sampling: cukup (Fs > 2f) dan under-sampled (Fs < 2f)
Fs_good = 100  # Sesuai Nyquist
Fs_bad = 40  # Kurang dari Nyquist

t_good = np.arange(0, 1, 1 / Fs_good)
t_bad = np.arange(0, 1, 1 / Fs_bad)

signal_good = np.sin(2 * np.pi * f_signal * t_good)
signal_bad = np.sin(2 * np.pi * f_signal * t_bad)

# Plot hasil
plt.figure(figsize=(12, 5))

plt.plot(
    t_cont, signal_cont, label="Sinyal Kontinu (30 Hz)", color="gray", linestyle="--"
)
plt.stem(
    t_good,
    signal_good,
    linefmt="g-",
    markerfmt="go",
    basefmt=" ",
    label="Sampling Cukup (Fs = 100 Hz)",
)
plt.stem(
    t_bad,
    signal_bad,
    linefmt="r-",
    markerfmt="ro",
    basefmt=" ",
    label="Under-sampling (Fs = 40 Hz)",
)

plt.title("Sampling Theorem - Efek Under-Sampling")
plt.xlabel("Waktu (s)")
plt.ylabel("Amplitudo")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
